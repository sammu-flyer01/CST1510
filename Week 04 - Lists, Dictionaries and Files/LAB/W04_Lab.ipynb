### D1. Create a list and index into it

### D2. Slicing

1. Create `values = [5, 10, 15, 20, 25, 30]`.
2. Use slicing to print: the middle four, the first two, and everything from position 3 to the end.

**Expected output:**
```
[10, 15, 20, 25]
[5, 10]
[20, 25, 30]
```

*Remember: the stop position is not included.*

````python
# D2
values = [5, 10, 15, 20, 25, 30]

# Middle four (items from index 1 up to 5)
print(values[1:5])

# First two (items from index 0 up to 2)
print(values[:2])

# From position 3 to the end
print(values[3:])
````

### D3. Change a list, then loop over it

### D6. List comprehensions

1. Create `values = [40, 55, 70, 25]`.
2. Using a list comprehension, make a new list with every value doubled, and print it.
3. Using a list comprehension with `if`, make a new list with only the values that are 50 or more, and print it.

**Expected output:**

````python
# D6
# 1. Create values
values = [40, 55, 70, 25]

# 2. List comprehension to double every value
doubled = [val * 2 for val in values]
print(doubled)

# 3. List comprehension with if to filter values 50 or more
filtered = [val for val in values if val >= 50]
print(filtered)
````

### D7. Create, read, add and update

1. Create `record = {"label": "R-001", "value": 87.0, "limit": 100.0}`.

### D8. Missing keys and `.get()`

1. Print `record.get("status")`.
2. Print `record.get("owner")`: this key does not exist.
3. Print `record.get("owner", "n/a")`.

**Expected output:** `OK`, `None`, `n/a`

````python
# D8
# 1. Print status using .get()
print(record.get("status"))

# 2. Print .get() for a missing key (returns None by default)
print(record.get("owner"))

# 3. Print .get() for a missing key with a default fallback value
print(record.get("owner", "n/a"))
````

### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```
label: R-001
value: 90.0
limit: 100.0
status: OK
```

````python
# D9
# a) Print only the keys
for key in record:
    print(key)

# b) Print only the values
for value in record.values():
    print(value)

# c) Print both as key: value
for key, value in record.items():
    print(f"{key}: {value}")
````

## Challenge · for a first

*These combine several ideas from this week. There are no step-by-step instructions.*

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
# 1. Create the list of dictionaries
records = [
    {"label": "R-001", "value": 87.0, "limit": 100.0},
    {"label": "R-002", "value": 120.0, "limit": 100.0},
    {"label": "R-003", "value": 45.0, "limit": 100.0},
]

# 2. Loop over records and evaluate values against their limits
for r in records:
    if r["value"] > r["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"
    
    print(f"{r['label']}: {status}")
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
# C2
def load_records(path):
    """Read label, value, limit lines from a file and return a list of dictionaries."""
    records = []
    try:
        with open(path) as f:
            for line in f:
                parts = line.strip().split(",")
                if len(parts) == 3:
                    label, value, limit = parts
                    records.append({
                        "label": label,
                        "value": float(value),
                        "limit": float(limit)
                    })
    except FileNotFoundError:
        print(f"File not found: {path}. Returning empty list.")
    
    return records

# Test it and print label and status as in D16
loaded_records = load_records("records_sample.txt")
for r in loaded_records:
    if r["value"] > r["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"
    print(f"{r['label']}: {status}")

# Test with missing file
load_records("nope.txt")
````
